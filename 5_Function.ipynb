# Price Taxer: Apply a 15% tax to prices over 100, and 5% tax to prices below 100

```python
def tax(a):
    if a>=100 :
        return a+a*(15/100) 
    else :
        return a+a*(5/100)
```

```python
tax(300)
```

```python
tax(90)
```

# String Palindrome Checker

```python
def palin():
    str1=input("Enter a string")
    str2=str1[::-1]
    if str1==str2:
        print("String is a palindrome")
    else:
        print("String is not a palindrome")    
```

```python
palin()
```

# Password Validator

```python
password = input("Enter a password having Min Lenght of 9 Char,upper and lower case,number and special character : ")
# len(password)
while True:
    if len(password) <= 8:
        print("Length must be greater than 8 char")
        password=input("Enter Password : ")
    if password.isupper() == True or password.islower() == True:
        print("Password must contain both upper and lower case")
        password=input("Enter Password : ")
    if any(char.isdigit()for char in password) == False:
        print("Password must contain number")
        password=input("Enter Password : ")
    if password.isalnum()==True:
        print("Password must contain special Character")
        password=input("Enter Password : ")
    print("Password verified")
    break
```

# Reverse each word in the sentence keeping its order same

```python
def reverse():
    sentence=input("Enter a sentence to reverse the words : ")
    result=""
    words=sentence.split()
    for word in words:
        result=result + word[::-1] + " "
    print(result.strip())       #Removes extra spaces
```

```python
reverse()
```
