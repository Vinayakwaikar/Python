# Q) Prime Number Logic

```python
number=int(input("Enter Number"))
if number>=2:
    is_prime=True
    
    for i in range(2,number):
        if number % i ==0:
            is_prime=False
            break
    
    if is_prime:
            print("Number is prime")
    else:
            print("Number is not prime")

else :
    print("Enter number should greater than or equal to 2")
```

# Q)FizzBuzz Program:
Print numbers from 1 to 50. For multiples of 3, print "Fizz", for multiples of 5 print "Buzz",
•and
for multiples of both, print "FizzBuzz".

```python
for i in range(1,51):
    
    if i%3 == 0 and i%5==0:
        print("FizzBuzz",i)
    elif i%3==0:
        print("Fizz", i)
    elif i%5==0:
        print("Buzz",i)
    else:
        print(i)
   
```
