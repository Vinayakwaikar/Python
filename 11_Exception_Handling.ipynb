# Custom Exceptions:

```python
class InsufficientFundsError(Exception):
    pass

def withdraw(balance,amount):
    if amount<balance:
        balance=balance-amount
        print('₹',amount,'debited from the account \n Your balance is ',balance)
    elif amount>balance:
        raise InsufficientFundsError('Insufficient balance')
        
```

```python
withdraw(1000,500)
```

```python
withdraw(2000,3000)
```

# Retrying Logic:

```python
password='Vinayak@56'
attempts=0
while attempts<3:
    try:
        username=input("Enter password")
        if username==password:
            print("Logined successfully")
            break
        else:
            attempts=attempts+1
            print(attempts, "attempts used out of 3");

    except Exception as e:
        print(e)
if attempts==3:
    print('Account Locked')
```
